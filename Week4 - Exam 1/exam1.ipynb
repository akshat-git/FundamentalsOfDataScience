```python
import pandas as pd

# Load the data

MOMA = pd.read_csv(r'Artworks MOMA-1-1-1.csv', encoding='ISO-8859-1')

```

```python
# Remove rows with missing values in the 'Title' column 

MOMA_cleaned = MOMA.dropna(subset=['Title'])
MOMA_cleaned = MOMA_cleaned.dropna(subset=['Artist'])
MOMA_cleaned = MOMA_cleaned.dropna(subset=['DateAcquired'])
total_rows = len(MOMA_cleaned)
total_rows

```

```python
photographs = MOMA_cleaned[MOMA_cleaned['Department'] == 'Photography']
photograph_count = len(photographs)
photograph_count
```

```python
percent_photographs = (photograph_count / total_rows) * 100
percent_photographs
```

```python
untitled_count = MOMA_cleaned[MOMA_cleaned['Title'].str.contains('untitled', case=False, na=False)]['Title'].nunique()
untitled_count
```

```python
paper_count = MOMA_cleaned[MOMA_cleaned['Medium'].str.contains('paper', case=False, na=False)].shape[0]

canvas_count = MOMA_cleaned[MOMA_cleaned['Medium'].str.contains('canvas', case=False, na=False)].shape[0]

ratio_paper_to_canvas = paper_count / canvas_count
ratio_paper_to_canvas
```
