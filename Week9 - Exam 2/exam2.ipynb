```python
from datascience import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
NUM_SIMS = 1000
sample_size = 300
def test_statistic(world_pop, observed_sample):
    return sum(abs(world_pop - observed_sample))
```

```python
world_pop = make_array(0.3, 0.4, 0.25, 0.05)

observed_sample = make_array(0.36, 0.45, 0.09, 0.10)

OBS_STAT = test_statistic(world_pop, observed_sample)


simulated_stats = make_array() 

for i in np.arange(NUM_SIMS):

    one_sample = sample_proportions(sample_size, world_pop)

    test_stat = test_statistic(world_pop, one_sample)

    simulated_stats = np.append(simulated_stats, test_stat)

p_value = np.count_nonzero(simulated_stats >= OBS_STAT) / NUM_SIMS
 
p_value
```
