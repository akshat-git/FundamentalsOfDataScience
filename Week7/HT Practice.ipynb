```python
sampled_ages = Table.read_table("age-1-1.csv")

gender_col = "gender" if "gender" in sampled_ages.labels else "Gender"
num_females = np.count_nonzero(np.char.lower(sampled_ages.column(gender_col).astype(str)) == "female")

num_females
```

```python
genders = np.char.lower(sampled_ages.column(gender_col).astype(str))
ages = sampled_ages.column("Age").astype(float)

male_age_avg = np.mean(ages[genders == "male"])
female_age_avg = np.mean(ages[genders == "female"])

observed_value = male_age_avg - female_age_avg
print("Observed value:", observed_value)
print("Rounded (a.bc):", f"{observed_value:.2f}")
```

```python
def simulate_one_statistic():
	shuffled_genders = np.random.permutation(genders)  # null: gender unrelated to age
	male_mean = np.mean(ages[shuffled_genders == "male"])
	female_mean = np.mean(ages[shuffled_genders == "female"])
	return male_mean - female_mean  # signed difference in means


# Simulate 4,000 values
num_simulations = 4000
simulated_stats = np.array([simulate_one_statistic() for _ in range(num_simulations)])

# Visualize
plt.figure(figsize=(8, 5))
plt.hist(simulated_stats, bins=30, edgecolor="white")
plt.axvline(np.mean(simulated_stats), color="orange", linewidth=2, label="Simulated mean")
plt.axvline(male_age_avg - female_age_avg, color="red", linestyle="--", linewidth=2, label="Observed diff")
plt.xlabel("Simulated (male mean - female mean)")
plt.ylabel("Count")
plt.title("Null Distribution of Difference in Mean Ages")
plt.legend()
output_path = "simulations_age_diff.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")
print("Center (simulated mean):", np.mean(simulated_stats))
```

```python
# Empirical (two-sided) p-value from the simulation
p_value = np.count_nonzero(np.abs(simulated_stats) >= observed_value) / num_simulations

print("p-value is about", f"{p_value:.2f}".replace("0.", "."))
```
