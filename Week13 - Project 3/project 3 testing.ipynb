```python
import pandas as pd
from scipy.spatial.distance import euclidean

# Load the CSV file
df = pd.read_csv('movies-1-1.csv')

# Select the rows for the given titles
row1 = df[df['Title'] == "10 things i hate about you"].iloc[0, 5:]
row2 = df[df['Title'] == "sister act"].iloc[0, 5:]

# Calculate Euclidean distance
distance = euclidean(row1, row2)
print("Euclidean distance:", distance)
```
